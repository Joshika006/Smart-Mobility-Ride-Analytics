```python
# Basic probability of each ride category

ride_probability = df["Ride_Category"].value_counts(normalize=True)

print("Probability of each ride category:")
print(ride_probability)
```

```python
# Probability of Peak and Non-Peak rides

time_probability = df["Ride_Time"].value_counts(normalize=True)

print("Probability of each ride time:")
print(time_probability)
```

```python
# Conditional probability: Premium given Peak

premium_peak = ((df["Ride_Category"] == "Premium") & 
                (df["Ride_Time"] == "Peak")).sum()

peak_total = (df["Ride_Time"] == "Peak").sum()

prob_premium_given_peak = premium_peak / peak_total

print("P(Premium | Peak):", prob_premium_given_peak)
```

```python
# Bayes' theorem

surge = df["Surge_Multiplier"] > 1

premium = df["Ride_Category"] == "Premium"

P_premium = premium.mean()
P_surge = surge.mean()
P_surge_given_premium = surge[premium].mean()

P_premium_given_surge = (P_surge_given_premium * P_premium) / P_surge

print("P(Premium | Surge):", P_premium_given_surge)
```

```python
p_economy = (df["Ride_Category"] == "Economy").mean()

print("Probability of Economy:", p_economy)
```

```python
from scipy.stats import binom

n = 10
k = 5

binomial_probability = binom.pmf(k, n, p_economy)

print("Probability of exactly 5 Economy rides:", binomial_probability)
```

```python
from scipy.stats import norm

fare_mean = df["Fare_Amount"].mean()
fare_std = df["Fare_Amount"].std()

print("Mean Fare:", fare_mean)
print("Standard Deviation:", fare_std)
```

```python
prob_fare_below_300 = norm.cdf(300, fare_mean, fare_std)

print("Probability of Fare < 300:", prob_fare_below_300)
```

```python
from scipy.stats import uniform

min_distance = df["Trip_Distance"].min()
max_distance = df["Trip_Distance"].max()

prob_distance = uniform.cdf(
    20,
    loc=min_distance,
    scale=max_distance - min_distance
)

print("Probability of Trip Distance <= 20:", prob_distance)
```

```python
import numpy as np

fare = df["Fare_Amount"].to_numpy()

bootstrap_means = []

for i in range(1000):
    sample = np.random.choice(fare, size=len(fare), replace=True)
    bootstrap_means.append(sample.mean())

print("Bootstrap Mean:", np.mean(bootstrap_means))
```

```python
lower = np.percentile(bootstrap_means, 2.5)
upper = np.percentile(bootstrap_means, 97.5)

print("95% Confidence Interval:")
print("Lower:", lower)
print("Upper:", upper)
```

```python
from scipy.stats import ttest_ind

premium_fare = df[df["Ride_Category"] == "Premium"]["Fare_Amount"]
economy_fare = df[df["Ride_Category"] == "Economy"]["Fare_Amount"]

t_stat, p_value = ttest_ind(premium_fare, economy_fare, equal_var=False)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

```python
from scipy.stats import chi2_contingency

table = pd.crosstab(df["Ride_Category"], df["Ride_Time"])

chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-Square Statistic:", chi2)
print("P-Value:", p_value)
```

```python
from scipy.stats import f

premium_var = premium_fare.var()
economy_var = economy_fare.var()

F = premium_var / economy_var

df1 = len(premium_fare) - 1
df2 = len(economy_fare) - 1

p_value = 2 * min(
    f.cdf(F, df1, df2),
    1 - f.cdf(F, df1, df2)
)

print("F-Statistic:", F)
print("P-Value:", p_value)
```
